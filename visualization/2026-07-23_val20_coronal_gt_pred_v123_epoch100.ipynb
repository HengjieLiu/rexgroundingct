# ReXGroundingCT Val20 Coronal GT/Prediction Review

First 20 cases from the reshuffled validation order. Predictions are from `v123_opt_poscrop_emptyloss`, epoch 100. Each patient figure uses coronal radiology display: top superior, bottom inferior, screen left patient right, screen right patient left.

```python
import os
from pathlib import Path
import sys

REPO_ROOT = Path.cwd()
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))

from visualization.rex_val20_coronal_viz import DEFAULT_OUTPUT_DIR, run_visualization

LIMIT = int(os.environ.get("REX_VIZ_LIMIT", "20"))
OUTPUT_DIR = Path(os.environ.get("REX_VIZ_OUTPUT_DIR", DEFAULT_OUTPUT_DIR))
WINDOW_CENTER = -600.0
WINDOW_WIDTH = 1500.0
OUTPUT_DIR
```

```python
result = run_visualization(
    limit=LIMIT,
    output_dir=OUTPUT_DIR,
    window_center=WINDOW_CENTER,
    window_width=WINDOW_WIDTH,
)
summary_df = result.summary_df
component_df = result.component_df
print(f"output_dir={result.output_dir}")
print(f"findings={len(summary_df)} components={len(component_df)} figures={len(result.figure_paths)}")
print(f"manifest={result.manifest_path}")
```
