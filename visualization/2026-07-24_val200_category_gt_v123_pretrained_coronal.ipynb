# ReXGroundingCT Val200 Category GT/V123/Pretrained Review

All 200 cases from the reshuffled validation order. Figures are grouped by patient and MICCAI category. Each figure compares GT contours, v123 epoch-100 predictions, and pretrained VoxTell v1.1 predictions on GT-component center-of-mass coronal slices.

```python
import os
from pathlib import Path
import sys

REPO_ROOT = Path.cwd()
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))

from visualization.rex_val20_coronal_viz import (
    DEFAULT_VAL200_CATEGORY_OUTPUT_DIR,
    run_visualization,
    val200_category_prediction_specs,
)

LIMIT_ENV = os.environ.get("REX_VIZ_LIMIT", "0")
LIMIT = None if LIMIT_ENV in {"", "0", "all", "ALL"} else int(LIMIT_ENV)
START_INDEX = int(os.environ.get("REX_VIZ_START_INDEX", "0"))
OUTPUT_DIR = Path(os.environ.get("REX_VIZ_OUTPUT_DIR", DEFAULT_VAL200_CATEGORY_OUTPUT_DIR))
WINDOW_CENTER = -600.0
WINDOW_WIDTH = 1500.0
DPI = int(os.environ.get("REX_VIZ_DPI", "110"))
OUTPUT_DIR
```

```python
result = run_visualization(
    limit=LIMIT,
    start_index=START_INDEX,
    output_dir=OUTPUT_DIR,
    prediction_specs=val200_category_prediction_specs(),
    category_grouped=True,
    window_center=WINDOW_CENTER,
    window_width=WINDOW_WIDTH,
    dpi=DPI,
)
summary_df = result.summary_df
component_df = result.component_df
print(f"output_dir={result.output_dir}")
print(f"findings={len(summary_df)} components={len(component_df)} figures={len(result.figure_paths)}")
print(f"manifest={result.manifest_path}")
```
